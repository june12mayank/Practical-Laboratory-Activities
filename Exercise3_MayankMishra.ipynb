## Exact Video Generation Prompt

> Create a cinematic 3-second video using the three provided reference images.
> Begin with the first original image displayed clearly. Transition smoothly
> into the second original image using a gentle crossfade. Finally, reveal the
> blended image as the concluding frame, demonstrating the visual combination
> of both source images. Use smooth camera movement, subtle zoom-in effects,
> seamless transitions, balanced lighting, and consistent visual quality.
> Maintain the original image content without introducing unrelated objects.
> The final result should look like a professional visual demonstration of
> image blending.

## Camera and Transition Instructions

| Setting | Instruction |
| --- | --- |
| Duration | 3 seconds |
| Aspect ratio | 16:9 |
| Camera movement | Gentle, continuous zoom-in |
| Transition 1 | Smooth crossfade from Image 1 to Image 2 (approximately 0.7–1.5 s) |
| Transition 2 | Smooth crossfade from Image 2 to the blended image (approximately 2.2–3.0 s) |
| Style | Cinematic, professional, balanced lighting |
| Content constraint | Preserve the source images; do not introduce unrelated objects |
| Audio | None specified |

This exercise demonstrates how a structured, multimodal prompt can specify the
subject, temporal order, camera motion, transitions, visual style, and content
constraints for a short AI video. The storyboard and local animatic make the
intended result reproducible and provide a clear reference for evaluating a
future Runway or Veo generation.
